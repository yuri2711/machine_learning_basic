```python
days_until_vacation = int(input("Введите количество дней до ближайшего отпуска: "))
weekends = 0
full_weeks = days_until_vacation // 7
weekends += full_weeks * 2
remaining_days = days_until_vacation % 7
for day in range(remaining_days):
    if day == 5 or day == 6:
        weekends += 1
print(f"Количество выходных дней до отпуска: {weekends}")
```

```python
length = int(input("Введите длину плитки шоколада: "))
width = int(input("Введите ширину плитки шоколада: "))
piece_size = int(input("Введите размер куска, который хотите отломить: "))
total_area = length * width
if piece_size <= total_area and (piece_size % length == 0 or piece_size % width == 0):
    print(True)
else:
    print(False)
```

```python
number = int(input("Введите целое положительное число: "))
val = [
    1000, 900, 500, 400,
    100, 90, 50, 40,
    10, 9, 5, 4,
    1
]
syms = [
    "M", "CM", "D", "CD",
    "C", "XC", "L", "XL",
    "X", "IX", "V", "IV",
    "I"
]

roman_numeral = ""
i = 0
while number > 0:
    for _ in range(number // val[i]):
        roman_numeral += syms[i]
        number -= val[i]
    i += 1
print(roman_numeral)
```

```python
#Проверка на положительные числа
value = input("Введите число. ")
if value == "":
    print(f'{value} -> False')

elif value[0] == '-':
    print(f'{value} -> False')
else:
    dot_found = False
    digits_found = False
    
    for char in value:
        if char.isdigit():
            digits_found = True
        elif char == '.':
            if dot_found:
                print(f'{value} -> False')
            dot_found = True
        else:
            print(f'{value} -> False')
    
    print(f'{value} -> {digits_found}')
```

```python
value = input("Введите число. ")
if value == "":
    print(f'{value} -> False')

elif value[0] != '-':
    print(f'{value} -> False')
else:
    dot_found = False
    digits_found = False
    negative_found = False
    
    for char in value:
        if char.isdigit():
            digits_found = True
        elif char == '.':
            if dot_found:
                print(f'{value} -> False')
            dot_found = True
        elif char == '-':
            if negative_found:
                print(f'{value} -> False')
            negative_found = True
        else:
            print(f'{value} -> False')
    
    print(f'{value} -> {digits_found}')
```
